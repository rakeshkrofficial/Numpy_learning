# Stacking

```python
# horizontal stacking

a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)

a4
```

```python
a5
```

```python
np.hstack((a4,a5))
```

```python
np.hstack((a4,a5,a4,a5))
```

```python
 # Vertical stacking
np.vstack((a4,a5))
```

```python
np.vstack((a4,a4,a5))
```
