# Iteration

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)

a1
```

```python
a2
```

```python
a3
```

```python
for i in a1:
    print(i)
```

```python
for i in a3:
    print(i)
```

```python
# 3D items print in seperate
```

```python
for i in np.nditer(a3):
    print(i)
```
