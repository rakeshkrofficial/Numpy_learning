```python
import numpy as np
```

# Creating Numpy Arrays


```python
# np.array

a = np.array([1,2,3,4,5])    # vector
print(type(a))
print(a)
```

#

```python
# 2D and 3D

b = np.array([[1,2,3,4],[5,4,3,2]])  # Metrix
b
```

```python
b = np.array([[[1,2],[4,5]],[[7,9],[4,3]]])  # Tensor
b
```

```python
# dtype

np.array([1,2,3,4],dtype = 'f')
```

```python
np.array([1,2,3,4],dtype = bool)
```

```python
np.array([1,2,3,4],dtype = complex)
```
